### Setting up ollama for LangChain with Model deepseek-r1 latest

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    base_url="http://localhost:11434",
    model="gemma4:latest",
    temperature=0.5
)
```

### Call the Ollama model

```python
response = llm.invoke("What is the capital of France?")
print(response.content)
```

### Streaming the response

```python
for chuck in llm.stream("What is the capital of France?"):
    print(chuck.content, end="",flush=True)
```

LangChain Chanining mechanism

```python
from langchain_core.prompts import ChatPromptTemplate
prompt_template = ChatPromptTemplate.from_template("What is the role of {subject} in software testing?")
chain =  prompt_template | llm
result = chain.invoke({"subject": "Human"})
print(result.content)
```

### Chaining with Parsers

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
prompt_template = ChatPromptTemplate.from_template("What is the role of {subject} in software testing?")
chain =  prompt_template | llm | StrOutputParser()  
result = chain.invoke({"subject": "Human"})
print(result)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import JsonOutputParser
prompt_template = ChatPromptTemplate.from_template("What is the role of {subject} in software testing? Give the answer in JSON format ONLY.")
chain =  prompt_template | llm | JsonOutputParser()  
result = chain.invoke({"subject": "Human"})
print(result)
```

### Chunking the text into smaller parts can be useful for processing large documents.Here's how you can do it using python:

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=50)
chunks = text_splitter.split_documents(documents)

for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}:")
    print(chunk.page_content)
    print("-" * 50)
```

### Building a simple agent with super simple tools

```python
from langchain.tools import tool
from langchain.agents import create_agent
from langchain.messages import HumanMessage

@tool
def add_numbers(a:int, b:int) -> int:
    "Adding two numbers together."
    return int(a) + int(b)

@tool
def subtract_numbers(a:int, b:int) -> int:
    "Subtracting two numbers."
    return int(a) - int(b)      

@tool
def multiply_numbers(a:int, b:int) -> int:
    "Multiplying two numbers."
    return int(a) * int(b)  

@tool
def divide_numbers(a:int, b:int) -> float:
    "Dividing two numbers."
    if int(b) == 0:
        raise ValueError("Cannot divide by zero.")
    return int(a) / int(b)

tools=[add_numbers, subtract_numbers, multiply_numbers, divide_numbers] 

#creating an AI Agent

agent = create_agent(
    tools= tools,
    model=llm
)
# With tool calls
query ="Whats the 100 less of 50?"
result =agent.invoke({"messages": [HumanMessage(content=query)]})
print(result["messages"][-1].content)

#without tool calls i.e direcly asking llm
query ="Whats the 100 less of 50?"
result =llm.invoke(query)
print(result)

# With tool calls
query ="Whats the 100 and 50 together"
result =agent.invoke({"messages": [HumanMessage(content=query)]})
print(result["messages"][-1].content)

# With tool calls
query ="Whats the 100 and 50 together and whats the 200 less of 50 and how about dobling the answers above"
result =agent.invoke({"messages": [HumanMessage(content=query)]})
print(result["messages"][-1].content)
```

```python
tools_invoked =[]

for message in result["messages"]:
    tools_calls =getattr(message, "tool_calls", None)

    if tools_calls:
        for tool_call in tools_calls:
            tool_name = tool_call["name"]
            tool_args = tool_call["args"]
            tools_invoked.append((tool_name, tool_args))

print("Tools invoked:",tools_invoked)           
```

Working with AI Agents

```python
from langchain_classic.agents import initialize_agent,AgentType

#Creating an AI Agent
agent = initialize_agent(
    tools=tools,
    llm=llm,
    agent=AgentType.STRUCTURED_CHAT_ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True,
)

query = "Extract the table data from http://eaapp.somee.com/Employee page and get me the average salary of the empolyee"

result = await agent.arun(query)
print(result)
```
